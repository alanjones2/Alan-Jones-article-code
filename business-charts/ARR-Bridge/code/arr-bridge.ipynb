```python
import pandas as pd
df= pd.DataFrame(
        [
            [10.00, 2.00, 1.50, 0.50, -1.00, -0.50, 0],
            [0,.0, 0, 0, 0, 0, 0],
        ],
        columns=[
                "Opening ARR",
                "New Business",
                "Expansion",
                "Price Increases",
                "Churn",
                "Contraction",
                "Closing ARR"],
        index = [["2025","2026"]]
)

df['Closing ARR']=df.sum(axis=1)
df = df.loc["2025"]
df
```

```python
values = [x for x in df.iloc[0]]
text_labels = [str(x) for x in values]

measure = ["relative"] * len(text_labels)
measure[0]="absolute"
measure[-1]="total"
categories = list(df.columns)

print(text_labels)
print(categories)
```

```python
import plotly.graph_objects as go

fig = go.Figure(go.Waterfall(
    measure = measure,
    x = categories,
    textposition = "outside",
    text = text_labels,
    y = values,
    
))

fig.update_layout(
        title = "Annual Recurring Revenue",
        template="plotly_white",
        width = 800,
        height = 400,
        margin=dict(l=25, r=25, t=25, b=25, pad=0)

)

fig.show()
```

### KPIs

arr_growth = (closing_ARR-open_ARR)/open_ARR*100

End_ARR


```python
closing_ARR = df['Closing ARR']
open_ARR = df['Opening ARR']
churn = df['Churn']
contraction = df['Contraction']
expansion  =df['Expansion']
price_increases = df['Price Increases']

arr_growth = (closing_ARR-open_ARR)/open_ARR*100
gross_revenue_retention = (open_ARR+churn+contraction)/open_ARR*100
net_revenue_retention = (open_ARR+expansion+price_increases+churn+contraction)/open_ARR*100

print("KPIs")
print(open_ARR, arr_growth,gross_revenue_retention,net_revenue_retention, closing_ARR)
```

```python

kpis = {
"ARR Growth": (closing_ARR-open_ARR)/open_ARR*100,
"Gross Revenue Retention": (open_ARR+churn+contraction)/open_ARR*100,
"Net Revenue Retention": (open_ARR+expansion+price_increases+churn+contraction)/open_ARR*100,
}
k = list(kpis.keys())
v = list(kpis.values())
for i in range(len(k)):
    print(k[i],":",v[i])
    

print(kpis)
```
